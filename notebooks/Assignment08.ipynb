## Exercise 2

<img src="../figures/lecture09/shield_curve.png" width="800"> 

The above figure is the "Shields curve" that determines the critical conditions for sediment motion on a bed, relating the nondimensionalized threshold fluid shear stress to the shear Reynolds number. It is a fundamental tool in river and coastal engineering, helping engineers predict when sediment will begin to move (incipient motion) and estimate scour depth and bedload transport. 

There are multiple parallel lines running from the bottom left to the top right, labeled “D = XX mm.” These represent particles of a given diameter but with different flow speeds, corresponding to a range of shear stress and Reynolds numbers—the greater the shear stress (faster flow), the larger the Reynolds number (more turbulent). Similarly, there are multiple parallel lines running from the bottom right to the top left, labeled “u = XX m/s.” These represent particles of varying sizes in a flow of a constant speed, corresponding to a range of shear stress and Reynolds numbers—the greater the shear stress (smaller particle), the smaller the Reynolds number (less turbulent).

The shaded band separates two zones: movement of sediment particles was observed in the zone above this shaded band, whereas no movement was observed in the zone underneath the shaded band.
**The shaded band therefore indicates initiation of motion.**

## Exercise 3

<img src="../figures/lecture10/shieldsno_bedload.png" width="600"> 

The Shields parameter is a fundamental tool for estimating the bedload transport. The above figure shows the dimensionless transport as a function of a **Shields parameter**. "Dimensionless" simply means multiplying or dividing other quantities in such a way that all units cancel out. 

The bedload transport (with units) can be calculated by multiplying the dimensionless transport $\Phi$ by $\sqrt{(s-1)gd_{50}^3}$, where $s$ is the specific density ($\rho_s/\rho_0$: sediment density divided by water density), $d_{50}$ is the median grain size. Therefore, the bedload transport has a unit of m$^3$/s/m (volumes of sand per second per metre width). 

The **Shields parameter $\theta'$** in the above figure:
> $\theta' = \frac{\tau_{b}}{(\rho_s-\rho_0)gd_{50}}$,
>
is a measure of the forcing on the sediment grains (drag and
lift) relative to the resisting force.
